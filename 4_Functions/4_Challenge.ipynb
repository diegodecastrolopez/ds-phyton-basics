For the first part of the assignment, we're going to get some practice taking something we've already written and translating it to a function.

1. Write a function that computes the factorial for an inputted number.

Write a function that determines whether or not an inputted number is prime, and then prints 'The number you inputted is a prime/ not a prime number.' depending on what your script finds (note that this means putting a `print` in front of the function when testing it will be redundant for this case). 

```python
def factorial(n):
    result = 1
    
    for i in range(1, n + 1):
        result *= i
    
    return result


# Testing the function
print(factorial(5))
```

```python
def is_prime(n):
    if n < 2:
        print("The number you inputted is not a prime number.")
        return
    
    for i in range(2, n):
        if n % i == 0:
            print("The number you inputted is not a prime number.")
            return
    
    print("The number you inputted is a prime number.")


# Testing the function
is_prime(13)
```

```python
4. def primes_up_to(n):
    primes = []
    
    for number in range(2, n + 1):
        is_prime = True
        
        for divisor in range(2, number):
            if number % divisor == 0:
                is_prime = False
                break
        
        if is_prime:
            primes.append(number)
    
    return primes


print(primes_up_to(10))
print(primes_up_to(20))
print(primes_up_to(5))
print(primes_up_to(30))
print(primes_up_to(50))
```

```python
5.def divisible_by(numbers, divisor):
    results = []
    
    for number in numbers:
        if number % divisor == 0:
            results.append("yes")
        else:
            results.append("no")
    
    return results


print(divisible_by([10, 25, 36, 12, 20], 5))
print(divisible_by([10, 15, 20], 5))
print(divisible_by([3, 6, 9, 10], 3))
print(divisible_by([7, 14, 21], 7))
print(divisible_by([8, 9, 10, 11], 2))
```

```python
6. 
def ends_with(strings, letter):
    results = []
    
    for string in strings:
        if string.endswith(letter):
            results.append(string)
    
    return results


print(ends_with(["I", "am", "in", "love", "with", "Python"], "n"))
print(ends_with(["cat", "dog", "bird", "fish"], "g"))
print(ends_with(["apple", "banana", "orange"], "e"))
print(ends_with(["run", "jump", "walk", "swim"], "m"))
print(ends_with(["hello", "world", "python"], "o"))
```

```python
def contains_substring(strings, substring):
    indices = []
    
    for index, string in enumerate(strings):
        if substring in string:
            indices.append(index)
    
    return indices


print(contains_substring(["This", "is", "an", "example"], "is"))
print(contains_substring(["apple", "banana", "orange"], "an"))
print(contains_substring(["hello", "world", "python"], "o"))
print(contains_substring(["cat", "dog", "bird"], "z"))
print(contains_substring(["testing", "this", "function"], "in"))
```
